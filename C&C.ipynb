## Trends in the regional funding distribution and the public participation in cultural heritage projects ##

In Italy, every region has different amount of funding and level of participation in cultural activities.
This project aims to discover how the public funding distributed in the last 10 years(2014-2024) is, and how the trend of visitors changes over time.

```python
import os
import glob

#function to extract the data from a list of csv files where only the year changes in the files name
#aim to have a table where the rows are in ascending chronological order and the columns are the italian regions

def transform_table(file):
    year = os.path.basename(file).split("_")[2]
    #detect header of table
    with open(file, "r", encoding = "latin-1") as f:
        for i,line in enumerate(f):
            if line.startswith("Regione"):
                header_row = i
                break
    visitors = pd.read_csv(file, skiprows = header_row, sep=";", encoding="latin-1")
    #one row and one column both named "Totale complessivo"
    visitors.rename(columns={"Totale complessivo":"Totale_regione"}, inplace= True) 
    # invert rows and columns
    visitors = visitors.set_index("Regione").T
    
    visitors.dropna(axis=1, how="all", inplace= True)
    visitors.dropna(axis=0, how="all", inplace= True)
    
    # rename the rows to add the year
    visitors.index = [f"{month}_{year}" for month in visitors.index]

    return visitors
```

```python
# apply to all visitatori files and create a unique table
files = glob.glob("data/musei_tavola5_*_visitatori.csv")
total_visitors = []
for f in files:
    total_visitors.append(transform_table(f))
    
# merge along rows
output_visitors = pd.concat(total_visitors)
output_visitors.to_csv("data/mic_visitors.csv", encoding="utf-8")
output_visitors
```

```python
import pandas as pd
import os
import glob

def transform_income(file):
    year = os.path.basename(file).split("_")[2]
    #detect header of table
    with open(file, "r", encoding = "latin-1") as f:
        for i,line in enumerate(f):
            if line.startswith("Regione"):
                header_row = i
                break
    money = pd.read_csv(file, skiprows = header_row, sep=";", encoding="latin-1")
    
    money.dropna(axis=1, how="all", inplace= True)
    money.dropna(axis=0, how="any", inplace= True)
    
    money = money[["Regione","Totale complessivo"]]
    money.rename(columns={"Totale complessivo": f"Introiti_{year}"}, inplace=True)

    return money
```

```python
# apply to all introiti files and create a unique table
tables_income = glob.glob("data/musei_tavola5_*_introiti.csv")
dfs = [transform_income(path) for path in tables_income]

# unisci tutti i DataFrame sulle regioni
output_income = dfs[0]
for d in dfs[1:]:
    output_income = output_income.merge(d, on="Regione", how="outer")
output_income = output_income[output_income["Regione"] != "Totale complessivo"]

output_income
```

```python
import pandas as pd

file_path = "data/open_coesione.csv"
df = pd.read_csv(file_path, dtype=str)

# FINANZ_TOTALE_PUBBLICO column
fin_col = "FINANZ_TOTALE_PUBBLICO"

# --- Convert financing to numeric ---
df[fin_col] = (
    df[fin_col].astype(str)
    .str.replace(r"[^\d,.\-]", "", regex=True)
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
)
df[fin_col] = pd.to_numeric(df[fin_col], errors="coerce")

#  Drop rows without province info 
df = df[df['DEN_PROVINCIA'].notna() & (df['DEN_PROVINCIA'].str.strip() != "")]

# Handle multi-region 
df['num_regions'] = df['DEN_REGIONE'].str.split(":::").str.len()
df_expanded = df.assign(DEN_REGIONE=df['DEN_REGIONE'].str.split(":::")).explode('DEN_REGIONE')
df_expanded['DEN_REGIONE'] = df_expanded['DEN_REGIONE'].str.strip()
df_expanded[fin_col] = df_expanded[fin_col] / df_expanded['num_regions']

# Normalize region names 
region_map = {
    "TRENTINO-ALTO ADIGE": "TRENTINO-ALTO ADIGE/SÜDTIROL",
    "TRENTINO ALTO ADIGE": "TRENTINO-ALTO ADIGE/SÜDTIROL",
    "VALLE D’AOSTA": "VALLE D'AOSTA",
    "FRIULI VENEZIA GIULIA": "FRIULI-VENEZIA GIULIA",
}
df_expanded['DEN_REGIONE'] = df_expanded['DEN_REGIONE'].replace(region_map)

# Drop non-regions
to_drop = ["PAESI EUROPEI", "AMBITO NAZIONALE"]
df_expanded = df_expanded[~df_expanded['DEN_REGIONE'].isin(to_drop)]

# Extract project start year 
df_expanded['OC_DATA_INIZIO_PROGETTO'] = pd.to_datetime(
    df_expanded['OC_DATA_INIZIO_PROGETTO'], errors="coerce"
)
df_expanded['YEAR'] = df_expanded['OC_DATA_INIZIO_PROGETTO'].dt.year.astype('Int64')
df_expanded = df_expanded[(df_expanded['YEAR'] >= 2014) & (df_expanded['YEAR'] <= 2024)]

# Aggregate funding by Region × Year
funding_matrix = df_expanded.groupby(['DEN_REGIONE', 'YEAR'])[fin_col].sum().reset_index()
```

```python
# pivot: regioni sulle righe, anni sulle colonne
funding_wide = funding_matrix.pivot(
    index="DEN_REGIONE",
    columns="YEAR",
    values=fin_col
)

# rinomina le colonne come Fondi_2014, Fondi_2015, ...
funding_wide.columns = [f"Fondi_{int(c)}" for c in funding_wide.columns]

# resetta l'indice per avere 'Regione' come colonna
funding_wide = funding_wide.reset_index().rename(columns={"DEN_REGIONE": "Regione"})

# merge orizzontale con la tabella degli introiti
merged = output_income.merge(funding_wide, on="Regione", how="outer")

# ordina le colonne nel formato Regione - Introiti_2014 - Fondi_2014 - ...
merged = merged.reindex(
    ["Regione"] +
    sorted([c for c in merged.columns if c != "Regione"], key=lambda x: (x[-4:], x))
, axis=1)

merged.to_csv("data/income&funding.csv", encoding="utf-8", index=False)
merged
```

```python
# fig.1.1

import pandas as pd
import plotly.graph_objects as go

# Load CSV
df = pd.read_csv("data/open_coesione.csv")


def parse_number(val):
    if pd.isna(val):
        return 0
    s = str(val).strip().replace("€", "").replace(" ", "")
    if "," in s and "." in s:
        s = s.replace(".", "").replace(",", ".")
    else:
        s = s.replace(",", ".")
    try:
        return float(s)
    except:
        return 0

def extract_year(date_str):
    if pd.isna(date_str):
        return None
    s = str(date_str)
    
    if len(s) >= 10 and s[4] == "-":
        return int(s[:4])
    
    if "/" in s and len(s.split("/")[-1]) == 4:
        return int(s.split("/")[-1])
    
    import re
    match = re.search(r"(19|20)\d{2}", s)
    return int(match.group(0)) if match else None


for col in ["FINANZ_TOTALE_PUBBLICO", "FINANZ_STATO_ESTERO", "FINANZ_PRIVATO"]:
    df[col] = df[col].apply(parse_number)


df["year"] = df["OC_DATA_INIZIO_PROGETTO"].apply(extract_year)


agg = df.groupby("year").agg({
    "FINANZ_TOTALE_PUBBLICO": "sum",
    "FINANZ_STATO_ESTERO": "sum",
    "FINANZ_PRIVATO": "sum"
}).reset_index()


agg["pubblico"] = agg["FINANZ_TOTALE_PUBBLICO"] / 1e6
agg["estero"]   = agg["FINANZ_STATO_ESTERO"] / 1e6
agg["privato"]  = agg["FINANZ_PRIVATO"] / 1e6

# --- Filter rules ---
agg = agg[(agg["year"] >= 1997) & (agg["year"] != 2025)]
agg = agg[(agg["pubblico"] > 0) | (agg["estero"] > 0) | (agg["privato"] > 0)]

fig = go.Figure()

fig.add_trace(go.Bar(
    x=agg["year"], y=agg["pubblico"],
    name="Pubblico", marker_color="#1976d2",
    hovertemplate="<b>%{x}</b><br>Pubblico: €%{y:.2f} M<extra></extra>"
))

fig.add_trace(go.Bar(
    x=agg["year"], y=agg["estero"],
    name="Estero", marker_color="#43a047",
    hovertemplate="<b>%{x}</b><br>Estero: €%{y:.2f} M<extra></extra>"
))

fig.add_trace(go.Bar(
    x=agg["year"], y=agg["privato"],
    name="Privato", marker_color="#ff7043",
    hovertemplate="<b>%{x}</b><br>Privato: €%{y:.2f} M<extra></extra>"
))


fig.update_layout(
    barmode="stack",
    xaxis=dict(title="Start Year", tickmode="linear", dtick=1, tickangle=-90),
    yaxis=dict(title="Funding (Million €)"),
    margin=dict(t=40, l=80, r=30, b=100),
    height=600,
    width=950
)
import plotly.io as pio
pio.renderers.default = "notebook" 

fig.show()
```

**Italian Public Funding of Cultural Heritage grouped by Regions** - fig.2.1

```python
# fig.2.1

import pandas as pd
import geopandas as gpd
import folium
from branca.colormap import linear

# --- 1. Helper function to normalize region names ---
def normalize_region_name(name: str) -> str:
    if not isinstance(name, str):
        return ""
    normalized = "".join(ch for ch in name.lower() if ch.isalnum())
    if "trentino" in normalized:
        normalized = "trentinoaltoadige"
    if "valle" in normalized:
        normalized = "valledaosta"
    return normalized

# Load the data 
csv_file = "data/open_coesione.csv"
geojson_file = "data/italy_regions.geojson"

df = pd.read_csv(csv_file, dtype=str)
gdf = gpd.read_file(geojson_file)

# Clean and process funding data 
df["FINANZ_TOTALE_PUBBLICO"] = (
    df["FINANZ_TOTALE_PUBBLICO"]
    .fillna("0")
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
    .astype(float)
)

df["region_norm"] = df["DEN_REGIONE"].apply(normalize_region_name)

# Aggregate funding per region
region_data = df.groupby("region_norm")["FINANZ_TOTALE_PUBBLICO"].sum().to_dict()

# Normalize names in GeoDataFrame and join funding data 
if "name" in gdf.columns:
    name_col = "name"
elif "reg_name" in gdf.columns:
    name_col = "reg_name"
else:
    print("Available columns in GeoJSON:", gdf.columns)
    raise KeyError("Could not find a region name column (expected 'name' or 'reg_name').")

gdf["region_norm"] = gdf[name_col].apply(normalize_region_name)
gdf["funding"] = gdf["region_norm"].map(region_data).fillna(0)


#  Create color scale 
max_funding = gdf["funding"].max()
colormap = linear.YlOrRd_09.scale(0, max_funding)
colormap.caption = "Total Funding (€)"

# Initialize map 
m = folium.Map(location=[41.87194, 12.56738], zoom_start=5, tiles="cartodbpositron")

#  Add GeoJSON layer with style + popups 
folium.GeoJson(
    gdf,
    style_function=lambda feature: {
        "fillColor": colormap(feature["properties"]["funding"]),
        "color": "white",
        "weight": 2,
        "dashArray": "3",
        "fillOpacity": 0.7,
    },
    highlight_function=lambda feature: {
        "weight": 5,
        "color": "#666",
        "dashArray": "",
        "fillOpacity": 0.9,
    },
    tooltip=folium.GeoJsonTooltip(
    fields=["reg_name", "funding"],
    aliases=["Region:", "Total Funding (€):"],
    localize=True
),
).add_to(m)

# Add legend
colormap.add_to(m)

m
```

```python
# fig.6.1/fig.6.2

import pandas as pd
import geopandas as gpd
import folium
from ipywidgets import interact, Dropdown
from IPython.display import display

#  Normalize names 
def normalize_region_name(name: str) -> str:
    if not isinstance(name, str):
        return ""
    normalized = "".join(ch for ch in name.lower() if ch.isalnum())
    if "trentino" in normalized:
        normalized = "trentinoaltoadige"
    if "valle" in normalized:
        normalized = "valledaosta"
    return normalized

# Load data 
df = pd.read_csv("data/mic_visitors.csv", dtype=str)
gdf = gpd.read_file("data/italy_regions.geojson")

# Keep only yearly totals rows
df = df[df["Unnamed: 0"].astype(str).str.contains("Totale_regione_", na=False)].copy()

# Extract Year column
df["Year"] = df["Unnamed: 0"].str.extract(r"Totale_regione_(\d{4})")
# If extraction failed for some rows, they will be NaN — convert to numeric
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")

# Define region columns (exclude Unnamed: 0 and Totale complessivo and Year) ---
numeric_cols = [c for c in df.columns if c not in ["Unnamed: 0", "Totale complessivo", "Year"]]

# Numeric cleaning function
def clean_numeric_series(s):
    # s is string series: remove dots used as thousands sep, replace comma decimal, strip spaces
    s_clean = s.fillna("0").astype(str).str.strip()
    # Remove any non-digit except comma and minus and dot (we will remove dots anyway)
    # First remove thousands dots
    s_clean = s_clean.str.replace(".", "", regex=False)
    # Replace comma decimal with dot (if present)
    s_clean = s_clean.str.replace(",", ".", regex=False)
    # Remove any stray non-numeric characters (except digits, dot, minus)
    s_clean = s_clean.str.replace(r"[^\d\.\-]", "", regex=True)
    # Convert to numeric coercing errors to NaN
    return pd.to_numeric(s_clean, errors="coerce")

# Apply cleaning to numeric columns
for col in numeric_cols:
    df[col] = clean_numeric_series(df[col])
    
# Ensure Totale complessivo exists and numeric
if "Totale complessivo" not in df.columns:
    raise KeyError("Column 'Totale complessivo' not found in dataframe columns.")
if not pd.api.types.is_numeric_dtype(df["Totale complessivo"]):
    df["Totale complessivo"] = clean_numeric_series(df["Totale complessivo"])
    
# Define region columns (exclude Unnamed: 0 and Totale complessivo and Year) ---
region_cols = [c for c in df.columns if c not in ["Unnamed: 0", "Totale complessivo", "Year"]]

# Reshape to long format (Year, Region, Visitors) 
long_df = df.melt(id_vars=["Year", "Totale complessivo"], var_name="Region", value_name="Visitors")

# Map Totale complessivo to each row explicitly and clean the column name
long_df = long_df.rename(columns={"Totale complessivo": "TotalYear"})

# Convert Visitors and TotalYear to numeric explicitly (again, defensive) ---
long_df["Visitors"] = pd.to_numeric(long_df["Visitors"], errors="coerce").fillna(0.0)
long_df["TotalYear"] = pd.to_numeric(long_df["TotalYear"], errors="coerce").fillna(0.0)

# Normalize region names ---
long_df["region_norm"] = long_df["Region"].apply(normalize_region_name)

# Compute Percentage relative to TotalYear
def safe_percentage(visitors, total):
    if pd.isna(total) or total == 0:
        return 0.0
    return (visitors / total) * 100.0

long_df["Percentage"] = long_df.apply(lambda r: safe_percentage(r["Visitors"], r["TotalYear"]), axis=1)

# Normalize GeoDataFrame
if "name" in gdf.columns:
    name_col = "name"
elif "reg_name" in gdf.columns:
    name_col = "reg_name"
else:
    raise KeyError("Could not find a region name column.")

gdf["region_norm"] = gdf[name_col].apply(normalize_region_name)

#  Fixed color and opacity 
fixed_color = "#3186cc"
fixed_opacity = 0.6

#  Circle size scaling 
max_visitors = long_df["Visitors"].max() if not long_df["Visitors"].isna().all() else 1.0
min_radius = 4
max_radius = 30

def scale_radius(visitors):
    """Scale visitors to circle radius with perceptual adjustment (square root scaling)."""
    if visitors <= 0 or pd.isna(visitors):
        return 0
    scaled = (visitors / max_visitors) ** 0.5
    return min_radius + scaled * (max_radius - min_radius)

# Draw map for a given year 
def draw_map(selected_year=2014):
    selected_year = int(selected_year)
    m = folium.Map(location=[41.87194, 12.56738], zoom_start=5.5, tiles="cartodbpositron")
    
    year_df = long_df[long_df["Year"] == selected_year].set_index("region_norm")
    
    if year_df.empty:
        folium.map.Marker(
            [41.87194, 12.56738],
            popup=f"No Totale_regione row found for year {selected_year}",
        ).add_to(m)
        return m
    
    for _, row in gdf.iterrows():
        region = row["region_norm"]
        if region not in year_df.index:
            continue
        visitors = year_df.loc[region, "Visitors"]
        if visitors == 0 or pd.isna(visitors):
            continue
        percentage = year_df.loc[region, "Percentage"]
        centroid = row.geometry.centroid
        
        folium.CircleMarker(
            location=[centroid.y, centroid.x],
            radius=scale_radius(visitors),
            color=fixed_color,
            fill=True,
            fill_color=fixed_color,
            fill_opacity=fixed_opacity,
            popup=f"<b>{row[name_col]}</b><br>"
                  f"Year: {selected_year}<br>"
                  f"Visitors: {int(visitors):,} ({percentage:.1f}%)"
        ).add_to(m)
    
    display(m)
   
# Dropdown 2014-2024
year_dropdown = Dropdown(options=list(range(2014, 2025)), value=2014, description="Year:")


# Dropdown 2016–2019 and 2022
#filtered_years = [y for y in [2016, 2017, 2018, 2019, 2022] if y in long_df["Year"].unique()]
#year_dropdown = Dropdown(options=sorted(filtered_years),value=filtered_years[0],description="Year:")
interact(draw_map, selected_year=year_dropdown);
```

```python
# fig.8

import folium
import pandas as pd
import geopandas as gpd
import numpy as np
import re

# Helper: Normalize region names 
def normalize_region_name(name):
    if not isinstance(name, str):
        return ''
    normalized = ''.join(ch for ch in name.lower() if ch.isalnum())
    if 'trentino' in normalized:
        normalized = 'trentinoaltoadige'
    if 'valle' in normalized:
        normalized = 'valledaosta'
    return normalized

# Load data
funding_df = pd.read_csv('data/open_coesione.csv')
visitor_df = pd.read_csv('data/mic_visitors.csv')
geo_df = gpd.read_file('data/italy_regions.geojson')

# Detect region name column in GeoJSON 
possible_name_cols = ['name', 'reg_name', 'REGIONE', 'DEN_REG', 'DENOMINAZI']
name_col = next((c for c in possible_name_cols if c in geo_df.columns), None)

if name_col is None:
    raise ValueError(f"Could not find a region name column in GeoJSON. Found columns: {geo_df.columns.tolist()}")

# Clean funding data 
def clean_number(x):
    if pd.isna(x):
        return 0
    x = str(x).replace('.', '').replace(',', '.')
    try:
        return float(x)
    except ValueError:
        return 0

funding_df['FINANZ_TOTALE_PUBBLICO'] = funding_df['FINANZ_TOTALE_PUBBLICO'].apply(clean_number)
funding_df['region_norm'] = funding_df['DEN_REGIONE'].apply(normalize_region_name)

funding_region_data = funding_df.groupby('region_norm')['FINANZ_TOTALE_PUBBLICO'].sum().to_dict()

# Add normalized names to GeoJSON 
geo_df['region_norm'] = geo_df[name_col].apply(normalize_region_name)

# Prepare visitor data 
long_data = []
for i, row in visitor_df.iterrows():
    year_match = None
    if isinstance(row['Unnamed: 0'], str):
        m = re.search(r'\d{4}', row['Unnamed: 0'])
        if m:
            year_match = int(m.group(0))
    for region, value in row.items():
        if region == 'Unnamed: 0':
            continue
        v = clean_number(value)
        long_data.append({
            'Year': year_match,
            'Region': region,
            'Visitors': v,
            'region_norm': normalize_region_name(region)
        })

long_df = pd.DataFrame(long_data)
years = sorted(long_df['Year'].dropna().unique())

# Create base map 
m = folium.Map(location=[41.87194, 12.56738], zoom_start=5.5, tiles='CartoDB positron')

# Funding choropleth layer 
folium.Choropleth(
    geo_data=geo_df,
    data=pd.DataFrame({'region_norm': list(funding_region_data.keys()), 'Funding': list(funding_region_data.values())}),
    columns=['region_norm', 'Funding'],
    key_on='feature.properties.region_norm',
    fill_color='YlOrRd',
    fill_opacity=0.6,
    line_opacity=0.5,
    legend_name='Funding (€)'
).add_to(m)

# Visitor circles for first year 
selected_year = years[0]
year_data = long_df[long_df['Year'] == selected_year]
total_visitors = year_data['Visitors'].sum()
max_visitors = year_data['Visitors'].max()

def scale_radius(v):
    min_radius, max_radius = 4, 25
    return 0 if v <= 0 else min_radius + np.sqrt(v / max_visitors) * (max_radius - min_radius)

for _, region in geo_df.iterrows():
    norm = region['region_norm']
    visitors = year_data.loc[year_data['region_norm'] == norm, 'Visitors'].sum()
    if visitors <= 0:
        continue
    centroid = region.geometry.centroid
    percentage = (visitors / total_visitors) * 100 if total_visitors > 0 else 0
    folium.CircleMarker(
        location=[centroid.y, centroid.x],
        radius=scale_radius(visitors),
        color='#3186cc',
        fill=True,
        fill_opacity=0.7,
        popup=f"<b>{region[name_col]}</b><br>Year: {selected_year}<br>Visitors: {int(visitors):,} ({percentage:.1f}%)"
    ).add_to(m)

# Add layer control 
folium.LayerControl().add_to(m)

# Display map in Jupyter 
m
```

```python
# fig.9

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

#  Load dataset 
file_path = "data/open_coesione.csv"
df = pd.read_csv(file_path, dtype=str)

#  Keep only relevant columns 
cols = [
    "CUP",
    "DATA_INIZIO_PREV_STUDIO_FATT",
    "OC_DATA_INIZIO_PROGETTO",
    "OC_DATA_FINE_PROGETTO_EFFETTIVA",
    "DEN_REGIONE",
    "FINANZ_TOTALE_PUBBLICO"
]
df = df[cols].copy()

# Helper: normalize region names 
def normalize_region_name(name: str) -> str:
    if not isinstance(name, str):
        return ""
    normalized = "".join(ch for ch in name.lower() if ch.isalnum())
    if "trentino" in normalized or "bolzano" in normalized or "altoadige" in normalized:
        return "trentinoaltoadige"
    elif "friuli" in normalized:
        return "friuliveneziagiulia"
    elif "valledaosta" in normalized or "aoste" in normalized:
        return "valledaosta"
    elif "emiliaromagna" in normalized:
        return "emiliaromagna"
    elif "lombardia" in normalized:
        return "lombardia"
    elif "piemonte" in normalized:
        return "piemonte"
    elif "veneto" in normalized:
        return "veneto"
    elif "liguria" in normalized:
        return "liguria"
    elif "toscana" in normalized:
        return "toscana"
    elif "umbria" in normalized:
        return "umbria"
    elif "marche" in normalized:
        return "marche"
    elif "lazio" in normalized:
        return "lazio"
    elif "abruzzo" in normalized:
        return "abruzzo"
    elif "molise" in normalized:
        return "molise"
    elif "campania" in normalized:
        return "campania"
    elif "puglia" in normalized:
        return "puglia"
    elif "basilicata" in normalized:
        return "basilicata"
    elif "calabria" in normalized:
        return "calabria"
    elif "sicilia" in normalized:
        return "sicilia"
    elif "sardegna" in normalized:
        return "sardegna"
    return normalized

#  Apply normalization 
df["REGION_KEY"] = df["DEN_REGIONE"].apply(normalize_region_name)

# Map normalized keys back to Italian names 
region_name_map = {
    "piemonte": "Piemonte",
    "valledaosta": "Valle d’Aosta",
    "lombardia": "Lombardia",
    "trentinoaltoadige": "Trentino-Alto Adige",
    "veneto": "Veneto",
    "friuliveneziagiulia": "Friuli-Venezia Giulia",
    "liguria": "Liguria",
    "emiliaromagna": "Emilia-Romagna",
    "toscana": "Toscana",
    "umbria": "Umbria",
    "marche": "Marche",
    "lazio": "Lazio",
    "abruzzo": "Abruzzo",
    "molise": "Molise",
    "campania": "Campania",
    "puglia": "Puglia",
    "basilicata": "Basilicata",
    "calabria": "Calabria",
    "sicilia": "Sicilia",
    "sardegna": "Sardegna"
}
df["DEN_REGIONE_CLEAN"] = df["REGION_KEY"].map(region_name_map)

#  Convert date columns to datetime 
for col in ["DATA_INIZIO_PREV_STUDIO_FATT", "OC_DATA_INIZIO_PROGETTO", "OC_DATA_FINE_PROGETTO_EFFETTIVA"]:
    df[col] = pd.to_datetime(df[col], errors='coerce')
    
#  FILTER for years 2014–2024 
mask = df['OC_DATA_INIZIO_PROGETTO'].dt.year.between(2014, 2024)
df = df[mask]

# Extract years 
df["year_prev"] = df["DATA_INIZIO_PREV_STUDIO_FATT"].dt.year
df["year_actual"] = df["OC_DATA_INIZIO_PROGETTO"].dt.year

# Split DataFrames: same vs delayed projects ---
same_year_df = df[df["year_prev"] == df["year_actual"]].copy()
delayed_df = df[df["year_prev"] != df["year_actual"]].copy()

print(f"Projects with same expected and actual start year: {len(same_year_df)}")
print(f"Projects with different expected and actual start year: {len(delayed_df)}")

# Calculate delay in years 
delayed_df["delay_years"] = delayed_df["year_actual"] - delayed_df["year_prev"]

#  Funding to numeric 
df["FINANZ_TOTALE_PUBBLICO"] = (
    df["FINANZ_TOTALE_PUBBLICO"]
    .str.replace(",", ".", regex=False)
    .astype(float, errors="ignore")
)

#  Average delay (overall) ---
average_delay = delayed_df["delay_years"].mean()
print(f"\nAverage project delay (years): {average_delay:.2f}")

#  Average delay by region (clean Italian names) 
avg_delay_by_region = (
    delayed_df.groupby("DEN_REGIONE_CLEAN")["delay_years"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

#  Funding allocation by region 
funding_by_region = (
    df.groupby("DEN_REGIONE_CLEAN")["FINANZ_TOTALE_PUBBLICO"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

#  Merge for summary 
region_summary = pd.merge(avg_delay_by_region, funding_by_region, on="DEN_REGIONE_CLEAN", how="left")

print("\nRegional summary (average delay + total funding):")
print(region_summary.head())

# --- Plotly bar chart ---
fig = px.bar(
    avg_delay_by_region,
    x="delay_years",
    y="DEN_REGIONE_CLEAN",
    orientation="h",
    text="delay_years",
    labels={"delay_years": "Average Delay (years)", "DEN_REGIONE_CLEAN": "Region"},
    title="fig.9: Average Delay of Projects per Region (2014–2024)",
    height=600
)

fig.update_traces(
    texttemplate="%{text:.2f}",
    textposition="outside",
    marker_color="teal"
)
fig.update_layout(yaxis=dict(autorange="reversed"))  # Largest delay on top

# --- Show and save as interactive HTML in one line ---
fig.show()  
#fig.write_html("graphs/regional_average_delay.html")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

try:
    # Load the dataset
    # The numbers use '.' for thousands and ',' for decimals.
    file_path = 'data/mic_income.csv'
    df = pd.read_csv(file_path, decimal=',', thousands='.', index_col=0)

    print("Data loaded successfully.")

    # --- Data Transformation ---
    
    # Extract the 'Year' from the index (e.g., 'Gennaio_2014' -> '2014')
    if not pd.api.types.is_datetime64_any_dtype(df.index):
        # Create a 'Year' column by splitting the index string
        df['Year'] = df.index.str.split('_').str[1]
    else:
        df['Year'] = df.index.year

    # Convert Year to integer for grouping
    df['Year'] = pd.to_numeric(df['Year'], errors='coerce')
    
    # Drop any rows where 'Year' could not be parsed
    df.dropna(subset=['Year'], inplace=True)
    
    # Cast Year to integer
    df['Year'] = df['Year'].astype(int)
    
    # Filter for the years 2014-2024, as requested
    df_filtered = df[(df['Year'] >= 2014) & (df['Year'] <= 2024)]

    # Group by the 'Year' column and sum up the monthly values
    df_yearly = df_filtered.groupby('Year').sum(numeric_only=True)

    # For the plot, we'll exclude 'Totale complessivo'
    # Its large value would make the other regions' lines flat and unreadable.
    if 'Totale complessivo' in df_yearly.columns:
        df_plot = df_yearly.drop(columns='Totale complessivo')
    else:
        df_plot = df_yearly
        
    print("Data successfully aggregated by year.")

    # --- Plotting ---
    
    plt.figure(figsize=(15, 10))
    
    # Plot each region as a separate line
    for region in df_plot.columns:
        plt.plot(df_plot.index, df_plot[region], label=region, marker='o', markersize=4)
        
    plt.title('Total Annual Income by Region (2014-2024)', fontsize=16)
    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Total Income (in Euro)', fontsize=12)
    
    # Ensure all years are shown as ticks on the x-axis
    plt.xticks(df_plot.index, rotation=45)
    
    # Format Y-axis to be more readable
    plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: format(int(x), ',')))
    
    plt.grid(True, linestyle='--', alpha=0.6)
    
    # Place the legend outside the main plot area
    plt.legend(bbox_to_anchor=(1.04, 1), loc='upper left', title="Regions")
    
    # Adjust layout to make space for the legend
    plt.tight_layout(rect=[0, 0, 0.85, 1]) 
    
    plot_filename = 'annual_income_by_region_2014_2024.png'
    plt.savefig(plot_filename)
    
    print(f"\nSuccessfully created and saved the graph as '{plot_filename}'.")

except Exception as e:
    print(f"An error occurred: {e}")
```
