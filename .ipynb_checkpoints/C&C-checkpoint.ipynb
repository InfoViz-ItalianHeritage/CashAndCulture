## Trends in the regional funding distribution and the public participation in cultural heritage projects ##

In Italy, every region has different amount of funding and level of participation in cultural activities.
This project aims to discover how the public funding distributed in the last 10 years(2014-2024) is, and how the trend of visitors changes over time.

```python
import os
import glob

#function to extract the data from a list of csv files where only the year changes in the files name
#aim to have a table where the rows are in ascending chronological order and the columns are the italian regions

def transform_table(file):
    year = os.path.basename(file).split("_")[2]
    #detect header of table
    with open(file, "r", encoding = "latin-1") as f:
        for i,line in enumerate(f):
            if line.startswith("Regione"):
                header_row = i
                break
    visitors = pd.read_csv(file, skiprows = header_row, sep=";", encoding="latin-1")
    #one row and one column both named "Totale complessivo"
    visitors.rename(columns={"Totale complessivo":"Totale_regione"}, inplace= True) 
    # invert rows and columns
    visitors = visitors.set_index("Regione").T
    
    visitors.dropna(axis=1, how="all", inplace= True)
    visitors.dropna(axis=0, how="all", inplace= True)
    
    # rename the rows to add the year
    visitors.index = [f"{month}_{year}" for month in visitors.index]

    return visitors
```

```python
# apply to all visitatori files and create a unique table
files = glob.glob("data/musei_tavola5_*_visitatori.csv")
total_visitors = []
for f in files:
    total_visitors.append(transform_table(f))
    
# merge along rows
output_visitors = pd.concat(total_visitors)
output_visitors.to_csv("data/mic_visitors.csv", encoding="utf-8")
output_visitors
```

```python
import pandas as pd
import os
import glob

def transform_income(file):
    year = os.path.basename(file).split("_")[2]
    #detect header of table
    with open(file, "r", encoding = "latin-1") as f:
        for i,line in enumerate(f):
            if line.startswith("Regione"):
                header_row = i
                break
    money = pd.read_csv(file, skiprows = header_row, sep=";", encoding="latin-1")
    
    money.dropna(axis=1, how="all", inplace= True)
    money.dropna(axis=0, how="any", inplace= True)
    
    money = money[["Regione","Totale complessivo"]]
    money.rename(columns={"Totale complessivo": f"Introiti_{year}"}, inplace=True)

    return money
```

```python
# apply to all introiti files and create a unique table
tables_income = glob.glob("data/musei_tavola5_*_introiti.csv")
dfs = [transform_income(path) for path in tables_income]

# unisci tutti i DataFrame sulle regioni
output_income = dfs[0]
for d in dfs[1:]:
    output_income = output_income.merge(d, on="Regione", how="outer")
output_income = output_income[output_income["Regione"] != "Totale complessivo"]

# converte tutte le colonne che iniziano con "Introiti_"
for col in [c for c in output_income.columns if c.startswith("Introiti_")]:
    output_income[col] = (
        output_income[col]
        .astype(str)
        .str.replace(".", "", regex=False)   # rimuove i punti delle migliaia
        .str.replace(",", ".", regex=False)  # converte la virgola in punto
    )
    output_income[col] = pd.to_numeric(output_income[col], errors="coerce")

output_income
```

```python
import pandas as pd

file_path = "data/open_coesione.csv"
df = pd.read_csv(file_path, dtype=str)

# FINANZ_TOTALE_PUBBLICO column
fin_col = "FINANZ_TOTALE_PUBBLICO"

# --- Convert financing to numeric ---
df[fin_col] = (
    df[fin_col].astype(str)
    .str.replace(r"[^\d,.\-]", "", regex=True)
    .str.replace(".", "", regex=False)   # rimuovi “.” separatore migliaia
    .str.replace(",", ".", regex=False) 
    
)
df[fin_col] = pd.to_numeric(df[fin_col], errors="coerce")

#  Drop rows without province info 
df = df[df['DEN_PROVINCIA'].notna() & (df['DEN_PROVINCIA'].str.strip() != "")]

# Handle multi-region 
df['num_regions'] = df['DEN_REGIONE'].str.split(":::").str.len()
df_expanded = df.assign(DEN_REGIONE=df['DEN_REGIONE'].str.split(":::")).explode('DEN_REGIONE')
df_expanded['DEN_REGIONE'] = df_expanded['DEN_REGIONE'].str.strip()
df_expanded[fin_col] = df_expanded[fin_col] / df_expanded['num_regions']

# Normalize region names 
region_map = {
    "TRENTINO-ALTO ADIGE": "TRENTINO-ALTO ADIGE/SÜDTIROL",
    "TRENTINO ALTO ADIGE": "TRENTINO-ALTO ADIGE/SÜDTIROL",
    "VALLE D’AOSTA": "VALLE D'AOSTA",
    "FRIULI VENEZIA GIULIA": "FRIULI-VENEZIA GIULIA",
    "EMILIA-ROMAGNA": "EMILIA ROMAGNA"
}
df_expanded['DEN_REGIONE'] = df_expanded['DEN_REGIONE'].replace(region_map)

# Drop non-regions
to_drop = ["PAESI EUROPEI", "AMBITO NAZIONALE"]
df_expanded = df_expanded[~df_expanded['DEN_REGIONE'].isin(to_drop)]

# Extract project start year 
df_expanded['OC_DATA_INIZIO_PROGETTO'] = pd.to_datetime(
    df_expanded['OC_DATA_INIZIO_PROGETTO'], errors="coerce"
)
df_expanded['YEAR'] = df_expanded['OC_DATA_INIZIO_PROGETTO'].dt.year.astype('Int64')
df_expanded = df_expanded[(df_expanded['YEAR'] >= 2014) & (df_expanded['YEAR'] <= 2024)]

# Aggregate funding by Region × Year
funding_matrix = df_expanded.groupby(['DEN_REGIONE', 'YEAR'])[fin_col].sum().reset_index()
```

```python
# pivot: regioni sulle righe, anni sulle colonne
funding_wide = funding_matrix.pivot(
    index="DEN_REGIONE",
    columns="YEAR",
    values=fin_col
)

# rinomina le colonne come Fondi_2014, Fondi_2015, ...
funding_wide.columns = [f"Fondi_{int(c)}" for c in funding_wide.columns]
funding_wide = funding_wide.round(2)       # due cifre decimali

# resetta l'indice per avere 'Regione' come colonna
funding_wide = funding_wide.reset_index().rename(columns={"DEN_REGIONE": "Regione"})

# merge orizzontale con la tabella degli introiti
merged = output_income.merge(funding_wide, on="Regione", how="outer")

# ordina le colonne nel formato Regione - Introiti_2014 - Fondi_2014 - ...
merged = merged.reindex(
    ["Regione"] +
    sorted([c for c in merged.columns if c != "Regione"], key=lambda x: (x[-4:], x))
, axis=1)

merged.to_csv("data/income&funding.csv", encoding="utf-8", index=False)
merged
```

#### **Italian Cultural Heritage Funding grouped by sources(public/private/foreign)** 

```python

import pandas as pd
import plotly.graph_objects as go

# Load CSV
df = pd.read_csv("data/open_coesione.csv")


def parse_number(val):
    if pd.isna(val):
        return 0
    s = str(val).strip().replace("€", "").replace(" ", "")
    if "," in s and "." in s:
        s = s.replace(".", "").replace(",", ".")
    else:
        s = s.replace(",", ".")
    try:
        return float(s)
    except:
        return 0

def extract_year(date_str):
    if pd.isna(date_str):
        return None
    s = str(date_str)
    
    if len(s) >= 10 and s[4] == "-":
        return int(s[:4])
    
    if "/" in s and len(s.split("/")[-1]) == 4:
        return int(s.split("/")[-1])
    
    import re
    match = re.search(r"(19|20)\d{2}", s)
    return int(match.group(0)) if match else None


for col in ["FINANZ_TOTALE_PUBBLICO", "FINANZ_STATO_ESTERO", "FINANZ_PRIVATO"]:
    df[col] = df[col].apply(parse_number)


df["year"] = df["OC_DATA_INIZIO_PROGETTO"].apply(extract_year)


agg = df.groupby("year").agg({
    "FINANZ_TOTALE_PUBBLICO": "sum",
    "FINANZ_STATO_ESTERO": "sum",
    "FINANZ_PRIVATO": "sum"
}).reset_index()


agg["pubblico"] = agg["FINANZ_TOTALE_PUBBLICO"] / 1e6
agg["estero"]   = agg["FINANZ_STATO_ESTERO"] / 1e6
agg["privato"]  = agg["FINANZ_PRIVATO"] / 1e6

# --- Filter rules ---
agg = agg[(agg["year"] >= 1997) & (agg["year"] != 2025)]
agg = agg[(agg["pubblico"] > 0) | (agg["estero"] > 0) | (agg["privato"] > 0)]

fig = go.Figure()

fig.add_trace(go.Bar(
    x=agg["year"], y=agg["pubblico"],
    name="Pubblico", marker_color="#1976d2",
    hovertemplate="<b>%{x}</b><br>Pubblico: €%{y:.2f} M<extra></extra>"
))

fig.add_trace(go.Bar(
    x=agg["year"], y=agg["estero"],
    name="Estero", marker_color="#43a047",
    hovertemplate="<b>%{x}</b><br>Estero: €%{y:.2f} M<extra></extra>"
))

fig.add_trace(go.Bar(
    x=agg["year"], y=agg["privato"],
    name="Privato", marker_color="#ff7043",
    hovertemplate="<b>%{x}</b><br>Privato: €%{y:.2f} M<extra></extra>"
))


fig.update_layout(
    barmode="stack",
    xaxis=dict(title="Start Year", tickmode="linear", dtick=1, tickangle=-90),
    yaxis=dict(title="Funding (Million €)"),
    margin=dict(t=40, l=80, r=30, b=100),
    height=600,
    width=950
)
import plotly.io as pio
pio.renderers.default = "notebook" 

fig.show()
```

#### **Italian Public Funding of Cultural Heritage grouped by Regions - Choropleth map**

```python

import pandas as pd
import geopandas as gpd
import folium
from branca.colormap import linear

# --- 1. Helper function to normalize region names ---
def normalize_region_name(name: str) -> str:
    if not isinstance(name, str):
        return ""
    normalized = "".join(ch for ch in name.lower() if ch.isalnum())
    if "trentino" in normalized:
        normalized = "trentinoaltoadige"
    if "valle" in normalized:
        normalized = "valledaosta"
    return normalized

# Load the data 
csv_file = "data/open_coesione.csv"
geojson_file = "data/italy_regions.geojson"

df = pd.read_csv(csv_file, dtype=str)
gdf = gpd.read_file(geojson_file)

# Clean and process funding data 
df["FINANZ_TOTALE_PUBBLICO"] = (
    df["FINANZ_TOTALE_PUBBLICO"]
    .fillna("0")
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
    .astype(float)
)

df["region_norm"] = df["DEN_REGIONE"].apply(normalize_region_name)

# Aggregate funding per region
region_data = df.groupby("region_norm")["FINANZ_TOTALE_PUBBLICO"].sum().to_dict()

# Normalize names in GeoDataFrame and join funding data 
if "name" in gdf.columns:
    name_col = "name"
elif "reg_name" in gdf.columns:
    name_col = "reg_name"
else:
    print("Available columns in GeoJSON:", gdf.columns)
    raise KeyError("Could not find a region name column (expected 'name' or 'reg_name').")

gdf["region_norm"] = gdf[name_col].apply(normalize_region_name)
gdf["funding"] = gdf["region_norm"].map(region_data).fillna(0)


#  Create color scale 
max_funding = gdf["funding"].max()
colormap = linear.YlOrRd_09.scale(0, max_funding)
colormap.caption = "Total Funding (€)"

# Initialize map 
m = folium.Map(location=[41.87194, 12.56738], zoom_start=5, tiles="cartodbpositron")

#  Add GeoJSON layer with style + popups 
folium.GeoJson(
    gdf,
    style_function=lambda feature: {
        "fillColor": colormap(feature["properties"]["funding"]),
        "color": "white",
        "weight": 2,
        "dashArray": "3",
        "fillOpacity": 0.7,
    },
    highlight_function=lambda feature: {
        "weight": 5,
        "color": "#666",
        "dashArray": "",
        "fillOpacity": 0.9,
    },
    tooltip=folium.GeoJsonTooltip(
    fields=["reg_name", "funding"],
    aliases=["Region:", "Total Funding (€):"],
    localize=True
),
).add_to(m)

# Add legend
colormap.add_to(m)

m
```

#### **Total Annual Income by Region (2014-2024) - Line graph**

```python
import pandas as pd
import matplotlib.pyplot as plt

try:
    # Load the dataset
    # The numbers use '.' for thousands and ',' for decimals.
    file_path = 'data/mic_income.csv'
    df = pd.read_csv(file_path, decimal=',', thousands='.', index_col=0)

    print("Data loaded successfully.")

    # --- Data Transformation ---
    
    # Extract the 'Year' from the index (e.g., 'Gennaio_2014' -> '2014')
    if not pd.api.types.is_datetime64_any_dtype(df.index):
        # Create a 'Year' column by splitting the index string
        df['Year'] = df.index.str.split('_').str[1]
    else:
        df['Year'] = df.index.year

    # Convert Year to integer for grouping
    df['Year'] = pd.to_numeric(df['Year'], errors='coerce')
    
    # Drop any rows where 'Year' could not be parsed
    df.dropna(subset=['Year'], inplace=True)
    
    # Cast Year to integer
    df['Year'] = df['Year'].astype(int)
    
    # Filter for the years 2014-2024, as requested
    df_filtered = df[(df['Year'] >= 2014) & (df['Year'] <= 2024)]

    # Group by the 'Year' column and sum up the monthly values
    df_yearly = df_filtered.groupby('Year').sum(numeric_only=True)

    # For the plot, we'll exclude 'Totale complessivo'
    # Its large value would make the other regions' lines flat and unreadable.
    if 'Totale complessivo' in df_yearly.columns:
        df_plot = df_yearly.drop(columns='Totale complessivo')
    else:
        df_plot = df_yearly
        
    print("Data successfully aggregated by year.")

    # --- Plotting ---
    
    plt.figure(figsize=(15, 10))
    
    # Plot each region as a separate line
    for region in df_plot.columns:
        plt.plot(df_plot.index, df_plot[region], label=region, marker='o', markersize=4)
        
    plt.title('Total Annual Income by Region (2014-2024)', fontsize=16)
    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Total Income (in Euro)', fontsize=12)
    
    # Ensure all years are shown as ticks on the x-axis
    plt.xticks(df_plot.index, rotation=45)
    
    # Format Y-axis to be more readable
    plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: format(int(x), ',')))
    
    plt.grid(True, linestyle='--', alpha=0.6)
    
    # Place the legend outside the main plot area
    plt.legend(bbox_to_anchor=(1.04, 1), loc='upper left', title="Regions")
    
    # Adjust layout to make space for the legend
    plt.tight_layout(rect=[0, 0, 0.85, 1]) 
    
    plot_filename = 'annual_income_by_region_2014_2024.png'
    plt.savefig(plot_filename)
    
    print(f"\nSuccessfully created and saved the graph as '{plot_filename}'.")

except Exception as e:
    print(f"An error occurred: {e}")
```

#### **Return Amount per Region (Introiti − Fondi) - Annotated Heatmap**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.ticker as mticker
import seaborn as sns

# Load CSV
df = pd.read_csv("data/income&funding.csv")

# Identify all years automatically
years = sorted({col.split("_")[1] for col in df.columns if "_" in col and col.split("_")[1].isdigit()})

# Compute (Introiti – Fondi) for every year
for y in years:
    df[f"Return_{y}"] = pd.to_numeric(df[f"Introiti_{y}"], errors="coerce") - \
                        pd.to_numeric(df[f"Fondi_{y}"], errors="coerce")

# Build pivot table for the Return columns
return_cols = [f"Return_{y}" for y in years]
pivot = df.set_index("Regione")[return_cols]

# Keep NaNs (do NOT fillna)
data = pivot.values
regions = pivot.index.tolist()

min_val = np.nanmin(data)
max_val = np.nanmax(data)

# Normalization centered at 0
norm = mcolors.TwoSlopeNorm(vmin=min_val, vcenter=0, vmax=max_val)

# Custom colormap: negatives red, positives blue, NaN = yellow
cmap = plt.cm.RdBu
cmap.set_bad(color="yellow")

plt.figure(figsize=(16, 10))
ax = sns.heatmap(
    pivot,
    cmap=cmap,
    linewidths=0.5,
    linecolor="gray",
    norm=norm,
    cbar_kws={
        'label': 'Return (Introiti − Fondi) €',
        'format': mticker.StrMethodFormatter('{x:,.0f}')
    },
    annot=pivot.map(lambda x: f"{x:,.0f}" if pd.notna(x) else ""),  # skip NaN annotations
    fmt="",
    annot_kws={"size": 8, "color": "black", "weight": "bold"}
)

ax.set_xticklabels(years, rotation=45, ha="right")
ax.set_yticklabels(regions, rotation=0)

plt.title("Annotated Heatmap — Return Amount (Introiti − Fondi)", fontsize=14, weight="bold")
plt.tight_layout()
plt.show()
```

#### **Regional Classification of TPL Efficiency per Year - Ranked barchart** 

```python
##to be added to website

import pandas as pd
import altair as alt

# 1. Load the Data
tpl_df = pd.read_csv("data/tpl_efficiency.csv")

# 2. Clean Region Names (using your map)
region_normalization_map = {
    "'Valle d\"'Aosta / Vallée d\"'Aoste'": "Valle d'Aosta",
    "Trentino Alto Adige / Südtirol": "Trentino-Alto Adige",
    "Emilia-Romagna": "Emilia-Romagna",
    "Friuli-Venezia Giulia": "Friuli-Venezia Giulia",
    "Abruzzo": "Abruzzo",
    "Basilicata": "Basilicata",
    "Calabria": "Calabria",
    "Campania": "Campania",
    "Lazio": "Lazio",
    "Liguria": "Liguria",
    "Lombardia": "Lombardia",
    "Marche": "Marche",
    "Molise": "Molise",
    "Piemonte": "Piemonte",
    "Puglia": "Puglia",
    "Sardegna": "Sardegna",
    "Sicilia": "Sicilia",
    "Toscana": "Toscana",
    "Umbria": "Umbria",
    "Veneto": "Veneto"
}

# Apply the cleaning
tpl_df['Regione'] = tpl_df['Regione'].map(region_normalization_map).fillna(tpl_df['Regione'])

# 3. Reshape Data (Melt) for Plotting
# We convert the wide format (columns Indice_2014, Indice_2015...) into a long format
tpl_cols = [col for col in tpl_df.columns if col.startswith('Indice_')]
tpl_melted = tpl_df.melt(
    id_vars=['Regione'], 
    value_vars=tpl_cols, 
    var_name='Year_Raw', 
    value_name='TPL_Index'
)
# Extract just the year number (e.g., 2014) from the column name
tpl_melted['Year'] = tpl_melted['Year_Raw'].str.replace('Indice_', '').astype(int)

# 4. Create the Interactive Ranking Chart
year_list = sorted(tpl_melted['Year'].unique())
year_slider = alt.binding_select(options=year_list, name='Select Year ')

# Selection for the slider
year_selector = alt.selection_point(
    fields=['Year'],
    bind=year_slider,
    value=[{'Year': year_list[0]}] # Initialize with the first year
)

# The Chart
chart = alt.Chart(tpl_melted).add_params(
    year_selector
).transform_filter(
    year_selector
).transform_window(
    # Calculate rank dynamically for the filtered data
    rank='rank(TPL_Index)',
    sort=[alt.SortField('TPL_Index', order='descending')] 
).encode(
    # X-axis: The value of the index
    x=alt.X('TPL_Index', title='Indice TPL (Efficienza)'),
    
    # Y-axis: Region, sorted by the Index value (Descending)
    y=alt.Y('Regione', sort='-x', title=None),
    
    # Color: Different color for each region
    color=alt.Color('Regione', legend=None, scale=alt.Scale(scheme='tableau20')),
    
    # Tooltip: Show details on hover
    tooltip=['Regione', 'TPL_Index', 'Year']
).mark_bar().properties(
    title='Classifica Regionale Indice TPL per Anno',
    width=600,
    height=500
)

# Add text labels to the end of the bars for clarity
text = chart.mark_text(
    align='left',
    baseline='middle',
    dx=3  # Nudges text to right so it doesn't overlap
).encode(
    text='TPL_Index'
)

# Combine bar chart and text
final_chart = (chart + text).interactive()

final_chart
```

#### **Strategic Matrix: Avg TPL vs. Avg Cult (Common Years) - Scatter Plot** 

```python
import pandas as pd
import altair as alt

# 1. Load Data
tpl_df = pd.read_csv("data/tpl_efficiency.csv")
cult_df = pd.read_csv("data/index_cult.csv")

# 2. Normalization Mapping
region_normalization_map = {
    "'Valle d\"'Aosta / Vallée d\"'Aoste'": "Valle d'Aosta",
    "Trentino Alto Adige / Südtirol": "Trentino-Alto Adige",
    "Emilia-Romagna": "Emilia-Romagna",
    "Friuli-Venezia Giulia": "Friuli-Venezia Giulia",
    "Abruzzo": "Abruzzo", "Basilicata": "Basilicata", "Calabria": "Calabria",
    "Campania": "Campania", "Lazio": "Lazio", "Liguria": "Liguria",
    "Lombardia": "Lombardia", "Marche": "Marche", "Molise": "Molise",
    "Piemonte": "Piemonte", "Puglia": "Puglia", "Sardegna": "Sardegna",
    "Sicilia": "Sicilia", "Toscana": "Toscana", "Umbria": "Umbria", "Veneto": "Veneto"
}

# Apply mapping
tpl_df['Regione'] = tpl_df['Regione'].map(region_normalization_map)
cult_df['Regione'] = cult_df['Regione'].map(region_normalization_map)

# 3. Transform Data (Melt)
tpl_cols = [col for col in tpl_df.columns if col.startswith('Indice_')]
tpl_melted = tpl_df.melt(id_vars=['Regione'], value_vars=tpl_cols, var_name='Year', value_name='TPL_Index')
tpl_melted['Year'] = tpl_melted['Year'].str.replace('Indice_', '').astype(int)

cult_cols = [col for col in cult_df.columns if col.startswith('Indice_')]
cult_melted = cult_df.melt(id_vars=['Regione'], value_vars=cult_cols, var_name='Year', value_name='Cult_Index')
cult_melted['Year'] = cult_melted['Year'].str.replace('Indice_', '').astype(int)

# 4. Merge DataFrames (Inner Join)
# Drops years where data is missing in either dataset to ensure fair comparison
merged_df = pd.merge(tpl_melted, cult_melted, on=['Regione', 'Year'], how='inner')

# 5. Calculate Averages
# Group by Region and calculate the mean of both indexes
average_df = merged_df.groupby('Regione')[['TPL_Index', 'Cult_Index']].mean().reset_index()

# 6. Plot the Average Matrix
base = alt.Chart(average_df).encode(
    x=alt.X('TPL_Index', title='Media Indice TPL (Efficienza Trasporti)'),
    y=alt.Y('Cult_Index', title='Media Indice Cult (Pubblico Interesse)'),
    color=alt.Color('Regione', legend=alt.Legend(title="Regione"), scale=alt.Scale(scheme='tableau20')),
    tooltip=[
        'Regione', 
        alt.Tooltip('TPL_Index', format='.0f', title='Media TPL'), 
        alt.Tooltip('Cult_Index', format='.2f', title='Media Cult')
    ]
).properties(
    title='Matrice Strategica: Media TPL vs. Media Cult (Anni Comuni)',
    width=800,
    height=500
)

# Create the dots
points = base.mark_circle(size=200)

# Create the labels
text = base.mark_text(
    align='left',
    baseline='middle',
    dx=10
).encode(
    text='Regione'
)

# Combine and make interactive 
chart = (points + text).interactive()

chart
```

#### **Horizontal Barchart of Total Public Funding Allocated to Regions (2014-2024)**

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns

#  Load dataset 
file_path = "data/open_coesione.csv"
df = pd.read_csv(file_path, dtype=str)

#  Keep only relevant columns 
cols = [
    "CUP",
    "DATA_INIZIO_PREV_STUDIO_FATT",
    "OC_DATA_INIZIO_PROGETTO",
    "OC_DATA_FINE_PROGETTO_EFFETTIVA",
    "DEN_REGIONE",
    "FINANZ_TOTALE_PUBBLICO"
]
df = df[cols].copy()

# Normalize region names 
def normalize_region_name(name: str) -> str:
    if not isinstance(name, str):
        return ""
    normalized = "".join(ch for ch in name.lower() if ch.isalnum())
    if "trentino" in normalized or "bolzano" in normalized or "altoadige" in normalized:
        return "trentinoaltoadige"
    elif "friuli" in normalized:
        return "friuliveneziagiulia"
    elif "valledaosta" in normalized or "aoste" in normalized:
        return "valledaosta"
    elif "emiliaromagna" in normalized:
        return "emiliaromagna"
    elif "lombardia" in normalized:
        return "lombardia"
    elif "piemonte" in normalized:
        return "piemonte"
    elif "veneto" in normalized:
        return "veneto"
    elif "liguria" in normalized:
        return "liguria"
    elif "toscana" in normalized:
        return "toscana"
    elif "umbria" in normalized:
        return "umbria"
    elif "marche" in normalized:
        return "marche"
    elif "lazio" in normalized:
        return "lazio"
    elif "abruzzo" in normalized:
        return "abruzzo"
    elif "molise" in normalized:
        return "molise"
    elif "campania" in normalized:
        return "campania"
    elif "puglia" in normalized:
        return "puglia"
    elif "basilicata" in normalized:
        return "basilicata"
    elif "calabria" in normalized:
        return "calabria"
    elif "sicilia" in normalized:
        return "sicilia"
    elif "sardegna" in normalized:
        return "sardegna"
    return normalized

#  Apply normalization 
df["REGION_KEY"] = df["DEN_REGIONE"].apply(normalize_region_name)

# Map normalized keys back to Italian names 
region_name_map = {
    "piemonte": "Piemonte",
    "valledaosta": "Valle d’Aosta",
    "lombardia": "Lombardia",
    "trentinoaltoadige": "Trentino-Alto Adige",
    "veneto": "Veneto",
    "friuliveneziagiulia": "Friuli-Venezia Giulia",
    "liguria": "Liguria",
    "emiliaromagna": "Emilia-Romagna",
    "toscana": "Toscana",
    "umbria": "Umbria",
    "marche": "Marche",
    "lazio": "Lazio",
    "abruzzo": "Abruzzo",
    "molise": "Molise",
    "campania": "Campania",
    "puglia": "Puglia",
    "basilicata": "Basilicata",
    "calabria": "Calabria",
    "sicilia": "Sicilia",
    "sardegna": "Sardegna"
}
df["DEN_REGIONE_CLEAN"] = df["REGION_KEY"].map(region_name_map)

# Convert date columns to datetime
date_cols = [
    "DATA_INIZIO_PREV_STUDIO_FATT",
    "OC_DATA_INIZIO_PROGETTO",
    "OC_DATA_FINE_PROGETTO_EFFETTIVA"
]
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")
    
#  FILTER for years 2014–2024 
mask = df['OC_DATA_INIZIO_PROGETTO'].dt.year.between(2014, 2024)
df = df[mask]

# Remove dots as thousand separators and replace comma with dot for decimals
df["FINANZ_TOTALE_PUBBLICO"] = (
    df["FINANZ_TOTALE_PUBBLICO"]
    .str.replace(".", "", regex=False)  # remove thousand separators
    .str.replace(",", ".", regex=False) # convert decimal comma to dot
)

# Convert to float
df["FINANZ_TOTALE_PUBBLICO"] = pd.to_numeric(
    df["FINANZ_TOTALE_PUBBLICO"], errors="coerce"
)

#  Funding allocation by region 
funding_by_region = (
    df.groupby("DEN_REGIONE_CLEAN")["FINANZ_TOTALE_PUBBLICO"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

# Visualization: Total funding by region 
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=funding_by_region,
    y="DEN_REGIONE_CLEAN",
    x="FINANZ_TOTALE_PUBBLICO",
    hue="DEN_REGIONE_CLEAN",
    palette="mako",
    legend=False
)


from matplotlib.ticker import FuncFormatter
# Custom formatter for K / M 
def millions_formatter(x, pos):
    if x >= 1e6:
        return f'{x*1e-6:.1f}M'  # millions
    elif x >= 1e3:
        return f'{x*1e-3:.0f}K'  # thousands
    else:
        return f'{x:.0f}'

ax.xaxis.set_major_formatter(FuncFormatter(millions_formatter))

plt.title("Total Public Funding Allocated to Regions (2014-2024)", fontsize=14, weight="bold")
plt.xlabel("Total Public Funding (€)")
plt.ylabel("Regions")
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()

"""
import mpld3
mpld3.save_html(plt.gcf(), "graphs/funding_barchart_10years.html")

"""
plt.show()
```
